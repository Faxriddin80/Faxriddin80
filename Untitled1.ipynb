```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

texts = [
    "suniy intellekt va mashina organish",
    "python dasturlash tili",
    "suniy intellekt algoritmlari",
    "futbol va sport yangiliklari",
    "basketbol sport turi",
    "python yordamida data tahlil",
    "tennis sport yangiliklari",
    "deep learning algoritmlari"
]


vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

print("Matnlar soni:", len(texts))
print("So'zlar soni:", len(vectorizer.get_feature_names_out()))


k = 3
model = KMeans(n_clusters=k, random_state=42)
model.fit(X)

labels = model.labels_

print("\nKlaster natijalari:")
for i, text in enumerate(texts):
    print(f"Klaster {labels[i]} : {text}")

terms = vectorizer.get_feature_names_out()
centers = model.cluster_centers_

for i in range(k):
    print(f"\nKlaster {i} asosiy so‘zlari:")
    top_words = centers[i].argsort()[-5:]

    for word in top_words:
        print(terms[word])


wcss = []

for i in range(1,8):
    model = KMeans(n_clusters=i, random_state=42)
    model.fit(X)
    wcss.append(model.inertia_)

plt.plot(range(1,8), wcss, marker="o")
plt.xlabel("Klasterlar soni")
plt.ylabel("WCSS")
plt.title("Optimal klaster sonini aniqlash")
plt.show()
```
